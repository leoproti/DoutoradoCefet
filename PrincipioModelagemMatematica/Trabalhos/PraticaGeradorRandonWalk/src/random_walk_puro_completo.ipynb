## 1. Passos gerais do Random Walk

A simulação do Random Walk segue estes passos principais:

1. Definir o número de passos `n_passos` e o número de caminhadas `n_caminhadas`.
2. Para cada caminhada, iniciar a posição em 0 (origem).
3. Em cada passo, gerar aleatoriamente `-1` ou `+1`.
4. Atualizar a posição somando este passo à posição atual.
5. Registrar a posição após cada passo para formar a trajetória.
6. No final, analisar o deslocamento final e a distribuição das posições.

## 2. Modelo teórico

No Random Walk simples em 1D, cada passo $X_i$ assume valores $+1$ ou $-1$ com probabilidade igual de 1/2.

$$S_N = \sum_{i=1}^{N} X_i, \quad X_i \in \{-1,+1\}, \quad P(X_i=+1)=P(X_i=-1)=\frac{1}{2}$$

O deslocamento final após $N$ passos é $S_N$.

A probabilidade exata de deslocamento final $d$ é dada por:

$$P_N(d) = \frac{1}{2^N} \binom{N}{\frac{N+d}{2}}$$

Isso é válido quando $N+d$ é par e $|d| \le N$; caso contrário, $P_N(d) = 0$.

Para grandes $N$, a distribuição de $S_N$ se aproxima de uma gaussiana:

$$P_N(d) \approx \frac{2}{\sqrt{2\pi N}} \exp\left(-\frac{d^2}{2N}\right)$$

## 3. Pseudocódigo

INÍCIO
    DEFINIR dim              // dimensão do espaço (1, 2, 3, ...)

        FIM PARA
        ADICIONAR TRAJETO à lista CAMINHADAS
        EXIBIR 'Caminhada X terminou em', última POSIÇÃO
        // --- FIM DE UMA CAMINHADA ---

A seguir implementamos este pseudocódigo em Python puro.

## 4. Implementação em Python puro

Nesta célula usamos apenas `random` para gerar os passos e `math` para cálculos auxiliares.

````python
import random
import math

# Parâmetros da simulação
dim = 1  # dimensão do espaço; use 1 para caminhada unidimensional
n_passos = 10000
n_caminhadas = 10
seed_base = 12345

# Função auxiliar para copiar a posição atual
def copiar_posicao(posicao):
    return posicao[:]

# Função que gera o valor do passo aleatório
def passo_aleatorio(rng):
    return -1 if rng.random() < 0.5 else 1

def simular_caminhada(dim, n_passos, seed):
    rng = random.Random(seed)
    posicao = [0] * dim
    trajeto = [copiar_posicao(posicao)]

    for passo in range(1, n_passos + 1):
        eixo = rng.randrange(dim)
        direcao = passo_aleatorio(rng)
        posicao[eixo] += direcao
        trajeto.append(copiar_posicao(posicao))

    return trajeto, copiar_posicao(posicao)

caminhadas = []
deslocamentos_finais = []

for i in range(1, n_caminhadas + 1):
    seed_atual = seed_base + i
    trajeto, pos_final = simular_caminhada(dim, n_passos, seed_atual)
    caminhadas.append(trajeto)
    deslocamentos_finais.append(pos_final)
    print(f'Caminhada {i} terminou em {pos_final}')

print()
print('Tamanho da primeira trajetória (n_passos + 1):', len(caminhadas[0]))
print('Deslocamentos finais:', deslocamentos_finais)
````

## 5. Análise dos deslocamentos finais

Calculamos a média e a variância dos deslocamentos finais para comparar com a teoria.

````python
media = sum(deslocamentos_finais) / len(deslocamentos_finais)
variancia = sum((d - media) ** 2 for d in deslocamentos_finais) / len(deslocamentos_finais)
sigma = math.sqrt(variancia)

print('Média dos deslocamentos finais:', media)
print('Variância dos deslocamentos finais:', variancia)
print('Desvio padrão dos deslocamentos finais:', sigma)

def p_n_d(n, d):
    if abs(d) > n or (n + d) % 2 != 0:
        return 0.0
    k = (n + d) // 2
    log_p = math.lgamma(n + 1) - math.lgamma(k + 1) - math.lgamma(n - k + 1) - n * math.log(2)
    return math.exp(log_p)

print()
print('Probabilidades teóricas para os deslocamentos finais observados:')
for d in deslocamentos_finais:
    print(f'  d = {d:>+5d}: P_N(d) = {p_n_d(n_passos, d):.3e}')
````
