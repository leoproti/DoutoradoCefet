# Random Walk e Curva Gaussiana

Este notebook apresenta um passeio aleatório (`random walk`) em uma dimensão e compara a distribuição das posições finais com uma curva gaussiana. O objetivo é explicar o conceito, mostrar o workflow de simulação e visualizar os resultados com gráficos.

## 1. Importar bibliotecas

Importar NumPy para geração de números aleatórios e cálculo de estatísticas, Matplotlib para visualização e Seaborn para melhorar os gráficos.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style='whitegrid', palette='deep', font_scale=1.1)
```

## 2. Definir parâmetros do random walk

Configurar o número de passos do passeio, a quantidade de simulações e a semente aleatória para garantir que os resultados sejam reprodutíveis.

```python
n_steps = 100
n_walks = 5000
seed = 2026

rng = np.random.default_rng(seed)

# Cada passo é +1 ou -1 com probabilidade igual
step_values = np.array([-1, 1])
```

## 3. Gerar o random walk

Criar a sequência cumulativa dos passos aleatórios e armazenar a trajetória do passeio. Cada linha de `walks` representa um passeio diferente.

```python
steps = rng.choice(step_values, size=(n_walks, n_steps))
walks = np.cumsum(steps, axis=1)

# Trajetória de um passeio exemplo
example_walk = walks[0]
final_positions = walks[:, -1]

example_walk[:10], final_positions.shape
```

## 4. Calcular densidade e curva gaussiana

A distribuição das posições finais tende à normal para um grande número de ensaios. Estimamos a média e o desvio padrão das posições finais e construímos a curva gaussiana associada.

```python
mean_final = np.mean(final_positions)
std_final = np.std(final_positions, ddof=1)

x_values = np.linspace(final_positions.min() - 2, final_positions.max() + 2, 300)
gaussian_curve = (1 / (std_final * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x_values - mean_final) / std_final) ** 2)

mean_final, std_final
```

## 5. Plotar random walk e curva gaussiana

Plotar a trajetória de um passeio de exemplo e a distribuição das posições finais de muitos passeios com a curva gaussiana sobreposta.

```python
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(14, 5))

# Gráfico 1: Trajetória do random walk
axes[0].plot(np.arange(1, n_steps + 1), example_walk, color='tab:blue', linewidth=1.5)
axes[0].axhline(0, color='black', linewidth=0.8, linestyle='--')
axes[0].set_title('Trajetória de um Random Walk (1D)')
axes[0].set_xlabel('Passo')
axes[0].set_ylabel('Posição')
axes[0].grid(True, alpha=0.4)

# Gráfico 2: Histograma das posições finais com ajuste gaussiano
sns.histplot(final_positions, bins=25, stat='density', color='tab:orange', alpha=0.6, ax=axes[1], edgecolor='black')
axes[1].plot(x_values, gaussian_curve, color='tab:blue', linewidth=2.2, label='Curva Gaussiana')
axes[1].axvline(mean_final, color='red', linestyle='--', label=f'Média = {mean_final:.2f}')
axes[1].set_title('Distribuição das posições finais e curva gaussiana')
axes[1].set_xlabel('Posição final')
axes[1].set_ylabel('Densidade')
axes[1].legend()
axes[1].grid(True, alpha=0.4)

plt.tight_layout()
plt.show()
```

### Interpretação dos resultados

- O passeio aleatório é construído somando passos discretos +1 ou -1.
- A trajetória de um único passeio mostra como a posição muda a cada passo.
- A distribuição das posições finais, após muitas simulações, aproxima-se de uma curva normal (Gaussiana), conforme prevê o teorema central do limite.
