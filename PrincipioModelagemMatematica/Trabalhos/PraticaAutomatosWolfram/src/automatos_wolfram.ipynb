# Automatos de Wolfram

Notebook base para explorar regras e gerar figuras para o relatorio.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

from wolfram_automato import initial_state, evolve, plot_grid
```

```python
rule = 30
width = 151
steps = 120

init = initial_state(width=width, mode='single', density=0.5, grouped=False, seed=42)
grid = evolve(rule=rule, init=init, steps=steps)

plot_grid(grid, rule=rule, show=True, save_fig=Path('graficos/regra30_single.png'))
```
